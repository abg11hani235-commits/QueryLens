# ContextQA - Phase 3: Baseline Model Inference

In this notebook, we load the pre-trained baseline model **`distilbert-base-uncased-distilled-squad`**.

> **Important Engineering Distinction**:
> - `distilbert-base-uncased-distilled-squad` is an **already fine-tuned** checkpoint provided by Hugging Face to serve as an out-of-the-box benchmark.
> - In **Phase 4**, we fine-tune `distilbert-base-uncased` (the raw pre-trained foundation model) ourselves on SQuAD to demonstrate the complete end-to-end training lifecycle.

```python
from src.inference import QAInferenceEngine
from src.utils import highlight_answer_in_context

# Initialize the inference engine with baseline checkpoint
engine = QAInferenceEngine(model_name_or_path="distilbert-base-uncased-distilled-squad")
```

## 1. Running Extractive QA on Benchmark Samples

```python
context = "Stanford University is a private research university in Stanford, California. It was founded in 1885 by Leland and Jane Stanford."
question = "When was Stanford University founded?"

result = engine.predict(question, context)
print(f"Question:   {question}")
print(f"Answer:     '{result['answer']}'")
print(f"Offsets:    [{result['start_char']}:{result['end_char']}]")
print(f"Raw Logits: {result['raw_score']}")
print(f"Confidence: {result['confidence'] * 100:.1f}%")

# Highlight in context
highlighted = highlight_answer_in_context(context, result['start_char'], result['end_char'], format_type="markdown")
print(f"\nHighlighted Context:\n{highlighted}")
```

## 2. Sliding Window Inference for Long Documents (>384 tokens)

```python
long_context = (
    "Artificial Intelligence has undergone exponential evolution over the last several decades. " * 35
    + " In 2017, the seminal paper Attention Is All You Need introduced the Transformer architecture, replacing recurrent models with self-attention mechanisms. "
    + " This revolutionized natural language understanding and machine translation worldwide. " * 25
)
long_question = "When was the Transformer architecture introduced?"

result_long = engine.predict(long_question, long_context)
print(f"Extracted Answer: '{result_long['answer']}'")
print(f"Confidence:       {result_long['confidence'] * 100:.1f}%")
print(f"Char Range:       [{result_long['start_char']}:{result_long['end_char']}]")
```
